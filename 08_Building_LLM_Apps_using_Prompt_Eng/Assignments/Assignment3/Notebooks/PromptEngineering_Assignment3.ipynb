# Extract Context from PDF

```python
def extract_text_from_pdf(pdf_path):
  """Extract text from a PDF file."""
  text = ""
  try:
    with open(pdf_path, 'rb') as file:
      reader = PyPDF2.PdfReader(file)
      for page in reader.pages:
        text += page.extract_text() + "\n"
    return text
  except Exception as e:
    print(f"Error extracting text from PDF: {e}")
    return None
```

```python
pdf_path = "company_annual_report.pdf"
document_text = extract_text_from_pdf(pdf_path)
print(f"Extracted {len(document_text)} characters from the PDF.")
```

## Enable Custom Knowledge to ChatGPT

```python
from openai import OpenAI
client = OpenAI(api_key=OPENAI_API_KEY)

def get_response(user_prompt, context, model="gpt-4o-mini"):
    messages = [
        {
            "role": "system",
            "content": (
                "You must answer ONLY using the following context. "
                "If the answer is not found in the context, reply with: "
                "\"I cannot answer from the provided context.\"\n\n"
                f"CONTEXT:\n{context}"
            )
        },
        {"role": "user", "content": user_prompt}
    ]

    response = client.chat.completions.create(
        model=model,
        messages=messages
    )

    return response.choices[0].message.content
```

```python
prompt1 ='What were the company’s total revenues in 2024?'
messages=[{"role":"user","content":prompt1}]
print(get_response(prompt1, document_text))
```

```python
prompt2 ='What are the key strategic initiatives mentioned in the report?'
messages=[{"role":"user","content":prompt2}]
print(get_response(prompt2, document_text))
```
