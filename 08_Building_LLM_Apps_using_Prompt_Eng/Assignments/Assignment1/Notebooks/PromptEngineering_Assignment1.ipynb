# Task 1: Classify Customer Reviews

You are provided with the following customer reviews for an online bookstore. Write
prompts to classify them into the categories: Delivery Issues, Product Quality, and
General Feedback.

## Reviews

1. "I ordered a book two weeks ago, and it still hasn't arrived. The tracking
information is not updating either."
2. "The book I received was damaged. The cover had scratches, and some pages
were torn."
3. "Your collection is fantastic! I found some rare books I've been searching for a
long time. Keep it up!"

## Objectives

*   Create a function that uses a chat model to classify each review into one of the
specified categories.
*   Ensure that the classification is based on the content of the Reviews.




# Response Function

temperature 0 makes sure to minimize the randomness and ensure a deterministic output

```python
def get_category(prompt, temperature=0, model="gpt-4o-mini"):
    sys_prompt = """Classify the following customer reviews into one of the following categories:
  1- Delivery Issues
  2- Product Quality
  3- General Feedback

  Generate the output following the below format:
  Review: <review>
  Category: <category>\n\n"""
    messages = [
        {"role": "system", "content": sys_prompt},
        {"role":"user","content":prompt}]
    response = openai.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature)

    return response.choices[0].message.content
```

```python
# Defining the Reviews
review1 = '''I ordered a book two weeks ago, and it still hasn't arrived. The tracking information is not updating either.'''
review2 = '''The book I received was damaged. The cover had scratches, and some pages were torn.'''
review3 = '''Your collection is fantastic! I found some rare books I've been searching for a long time. Keep it up!'''
#review4 = '''the last book I ordered was late to arrive and the cover was dented, i enjpyed reading it regardless'''
```

```python
# Generating the user prompt and the response as a function
def generate_category(review):
  user_prompt=f"Customer Reviews:{review}"
  return get_category(user_prompt)
```

```python
reviews = [review1, review2, review3]
for review in reviews:
  print(generate_category(review))
```

## Review:
"I recently bought a gaming mouse from your website. I use Windows 11, and the
mouse disconnects randomly during use. My order number is ORD123456. Please
assist."

## Required JSON Format:
{
"product": "",
"operating_system": "",

##Objective:


*   Write a Python function that generates the required structured JSON format
using the chat model.
*   Demonstrate how to parse and return the extracted details in the correct JSON
structure.

```python
# defining the review
review1 = '''I recently bought a gaming mouse from your website. I use Windows 11, and the mouse disconnects randomly during use. My order number is ORD123456. Please assist'''
```

```python
# define response function
def get_json(prompt, temperature=0, model="gpt-4o-mini"):
    sys_prompt = """
    Extract from the following review the following structured information following the specified format
    Required JSON Format:
    "product": "",
    "operating_system": "",
    "issue_description": "",
    "order_number": ""
    """
    messages = [
        {"role": "system", "content": sys_prompt},
        {"role":"user","content":prompt}]
    response = openai.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature)

    return response.choices[0].message.content
```

```python
# defining the user prompt
user_prompt = f'''Review: {review1} '''

print(get_json(user_prompt))
```

# Task 4: Create Product Names

## Objective

```python
# define response functions
def get_prod_low_temp(temperature=0, model="gpt-4o-mini"):
    sys_prompt = """Generate 5 creative product names for a smart home assistant."""
    messages = [
        {"role": "system", "content": sys_prompt}]
    response = openai.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature)

    return response.choices[0].message.content

# define response functions
def get_prod_high_temp(temperature=1, model="gpt-4o-mini"):
    sys_prompt = """Generate 5 creative product names for a smart home assistant."""
    messages = [
        {"role": "system", "content": sys_prompt}]
    response = openai.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature)

    return response.choices[0].message.content
```
