```python
from langchain_community.document_loaders import UnstructuredPDFLoader

from IPython.display import display, HTML
```

```python
# take around 3-5 minutes
loader = UnstructuredPDFLoader(
    "../data/attention_paper.pdf",
    mode="elements",
    strategy="hi_res",
    infer_table_structure=True,
    languages=["eng"],
    pdf_image_dpi=1200, # increased to see if parsing can be better
)

docs = loader.load()
```

```python
# get parts that were extracted
is_extracted_docs = [doc for doc in docs if doc.metadata.get("is_extracted") in ("partial", "true")]
```

```python
# get only tables
[doc.page_content for doc in docs if doc.metadata.get("category")=="Table"]

tables = [doc.metadata["text_as_html"] for doc in docs if doc.metadata.get("category")=="Table"]
```
