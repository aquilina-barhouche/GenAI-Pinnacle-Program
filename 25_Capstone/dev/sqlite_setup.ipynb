# SQLite Setup

## Imports

```python
import sqlite3
```

## Create Table

```python
connection = sqlite3.connect("../agaip/db/conversations.db")

cursor = connection.cursor()

cursor.execute("""
    CREATE TABLE IF NOT EXISTS sessions (
        guid TEXT PRIMARY KEY,
        created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
        user TEXT NOT NULL,
        title TEXT,
        messages TEXT
    )
""")

connection.commit()
```
