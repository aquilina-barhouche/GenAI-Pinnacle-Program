```python
from googleapiclient.discovery import build

def get_google_links(query, api_key, cx):
    # Build the service object
    service = build("customsearch", "v1", developerKey=api_key)
    
    # Execute the search request
    # 'num=10' ensures you get exactly the top 10 results
    res = service.cse().list(q=query, cx=cx, num=10).execute()
    
    # Extract only the links from the results
    links = [item['link'] for item in res.get('items', [])]
    
    return links

# Usage
MY_API_KEY = "INSERT_YOUR_API_KEY"
MY_CX = "INSERT_YOUR_CX_ID"
search_text = "best programming languages 2024"

top_10_links = get_google_links(search_text, MY_API_KEY, MY_CX)

for i, link in enumerate(top_10_links, 1):
    print(f"{i}. {link}")
```
